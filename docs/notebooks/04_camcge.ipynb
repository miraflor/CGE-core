## One published experiment

Experiment 1 changes foreign saving (`fsav`) to 500. CAMCGE keeps its own model-specific variable names and closure.

```python
windfall = base.scenario("Oil windfall")
windfall.set("fsav", None, 500)
windfall_result = windfall.solve()
windfall_result.summary()
```
