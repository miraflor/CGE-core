# Policy experiments

Solve the benchmark, create an independent scenario, apply an economic shock, solve again, and compare.

```python
try:
    import cge_core
except ImportError:
    %pip install -q "cge-core[solver]==0.7.0"
    from cge_core import install_solver
    install_solver()
```

```python
from cge_core import StandardCGE

base = StandardCGE.example().solve()
```

```python
reform = base.scenario("50% BRD tariff cut")
reform.tariff("BRD", change=-0.50)
result = reform.solve()
result.summary()
```

```python
comparison = result.compare(base)
comparison.query("component in ['Z', 'pq', 'Xp', 'E', 'M']")
```

A second scenario is independent of the first.

```python
resources = base.scenario("10% more capital")
resources.endowment("CAP", change=0.10)
resources.solve().compare(base).head()
```
