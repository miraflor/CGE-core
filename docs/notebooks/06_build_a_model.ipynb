## 1. Parse and validate `.cge.md`

```python
from importlib.resources import files
from cge_core.spec import parse_file, validate_document

path = files("cge_core.spec").joinpath("examples", "two_good_exchange.cge.md")
doc = validate_document(parse_file(path))
print(len(doc.equations), "equations")
print("shockable:", [g.names for g in doc.shockables])
```

Only fenced `cge` blocks are executable. Changing Markdown explanation does not change this AST.
