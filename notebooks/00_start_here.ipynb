## The recurring CGE idea

```text
benchmark data
      ↓
calibrate a baseline equilibrium
      ↓
change one or more exogenous assumptions
      ↓
solve the whole economy again
      ↓
compare scenario with baseline
```

A CGE shock changes an assumption. The model solves for the mutually consistent response of endogenous quantities and prices.

## 2. Inspect the two bundled Hosoe teaching economies

````python
import pandas as pd
from IPython.display import display
from cge_core import example_data

def read_sam(name):
    return pd.read_csv(example_data(name) / "param-sam-.csv", index_col=0)

simple_sam = read_sam("splcge")
standard_sam = read_sam("stdcge")

print("SIMPLE CGE — accounts:", list(simple_sam.index))
display(simple_sam)

print("\nSTANDARD CGE — accounts:", list(standard_sam.index))
display(standard_sam)
````

## 3. The conceptual jump

The **simple CGE** has two goods, two factors, and one household. There is no government, trade, tax system, or intermediate-input network.

The **standard CGE** adds intermediate inputs, government, taxes, imports and exports, Armington/CET structure, saving and investment, and the rest of the world.

````python
comparison = pd.DataFrame(
    [
        ["Hosoe simple", "splcge", 2, 2, "No", "No", "No", "No"],
        ["Hosoe standard", "stdcge", 2, 2, "Yes", "Yes", "Yes", "Yes"],
        ["IFPRI", "cge_core.ifpri", "Richer", "Richer", "Yes", "Yes", "Yes", "Yes"],
        ["CAMCGE", "cam/", 11, 3, "Yes", "Yes", "Yes", "Yes"],
    ],
    columns=["Model", "Location", "Goods/sectors", "Factors", "Government", "Trade", "Taxes", "Investment"],
)
display(comparison)
````
